#remove duplicates rows

```python
df1=df.drop_duplicates()
```

```python
df1
```

```python
#product    -convert ? mark to blank
df1['ProductName']=df1['ProductName'].str.replace('�',' ')
```

```python
df1
```

```python
#company name      -convert ? mark to blank
df1['CompanyName'][df1['CompanyName'].str.contains('�')]
df1['CompanyName']=df1['CompanyName'].str.replace('�',' ')
```

```python
df1
```

```python
#brand name      -convert ? mark to blank
df1['BrandName'].str.contains('�').sum()
```

```python
df1['BrandName']=df1['BrandName'].str.replace('�',' ')
```

```python
df1
```

#convert dtype of csfid

```python
df1['CSFId']=df1['CSFId'].fillna(0)
```

```python
df1['CSFId']=df1['CSFId'].astype('int32')
```

```python
df1.info()
```

#csf        -convert nan to data not available

```python
df1['CSF']=df1['CSF'].fillna('data not available')
```

```python
df1['CSF'].head(2)
```

#brandname -convert nan to data not available

```python
df1['BrandName']=df1['BrandName'].fillna('data not available')
df1['BrandName']
```

```python
df1.info()
```

#change dtype to datetime

```python
df1['InitialDateReported']=pd.to_datetime(df1['InitialDateReported'])
df1['MostRecentDateReported']=pd.to_datetime(df1['MostRecentDateReported'])
df1['ChemicalCreatedAt']=pd.to_datetime(df1['ChemicalCreatedAt'])
df1['ChemicalUpdatedAt']=pd.to_datetime(df1['ChemicalUpdatedAt'])
```

```python
df1.info()
```

#ChemicalDateRemoved            drop column

```python
df1=df1.drop('ChemicalDateRemoved',axis=1)
```
